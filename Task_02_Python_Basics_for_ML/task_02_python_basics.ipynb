Example 3: Float

```python
price = 1250.75
tax = 0.17

total = price + (price * tax)

print("Original Price:", price)
print("Total Price:", round(total, 2))
```

Example 4: Boolean

```python
age = 19
has_id = True

eligible = age >= 18 and has_id

print("Eligible:", eligible)
```

Example 5: List

```python
skills = ["Python", "GitHub", "NumPy"]

skills.append("Pandas")
skills.remove("GitHub")

print("Skills:", skills)
print("Total Skills:", len(skills))
```

Example 6: Tuple

```python
coordinates = (24.86, 67.01)

latitude, longitude = coordinates

print("Latitude:", latitude)
print("Longitude:", longitude)
```

Example 7: Dictionary

```python
student = {
    "Name": "Afnan Ali",
    "Age": 19,
    "City": "Kazi Ahmed"
}

for key, value in student.items():
    print(key, ":", value)
```
